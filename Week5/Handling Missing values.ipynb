```python
import pandas as pd
import numpy as np

data = {
    "CustomerID": [101,102,103,104,105],
    "Name": ["Rahul","Priya","Arjun","Sneha","Kiran"],
    "Age": [25, np.nan, 35, 28, np.nan],
    "Gender": ["Male","Female",np.nan,"Female","Male"],
    "City": ["Hyderabad","Bangalore","Chennai",np.nan,"Hyderabad"],
    "Salary": [45000,52000,np.nan,48000,55000]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Experiment 2: Handling Missing Values

import pandas as pd

# 1. Load the dataset
df = pd.read_csv("test.csv")

print("Original Dataset:")
print(df)

# ----------------------------------------------------
# 2. Detect missing values
# ----------------------------------------------------
print("\nMissing Values (True = Missing):")
print(df.isnull())

# ----------------------------------------------------
# 3. Count missing values in each column
# ----------------------------------------------------
print("\nMissing Values Count:")
print(df.isnull().sum())

# ----------------------------------------------------
# 4. Replace numerical missing values using Mean
# ----------------------------------------------------
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

for column in numerical_columns:
    df[column] = df[column].fillna(df[column].mean())

# ----------------------------------------------------
# 5. Replace categorical missing values using Mode
# ----------------------------------------------------
categorical_columns = df.select_dtypes(include=['object']).columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

# ----------------------------------------------------
# 6. Display dataset after preprocessing
# ----------------------------------------------------
print("\nDataset After Preprocessing:")
print(df)

# ----------------------------------------------------
# 7. Compare missing values before and after
# ----------------------------------------------------
print("\nRemaining Missing Values:")
print(df.isnull().sum())

# ----------------------------------------------------
# 8. Save the cleaned dataset
# ----------------------------------------------------
df.to_csv("cleaned_test.csv", index=False)

print("\nCleaned dataset saved as 'cleaned_test.csv'")
```
